## 三种 Memory 示例（控制发给模型的历史长度）

下面三个 block 分别演示：**窗口记忆**、**摘要+最近几轮**、**按 token 数截断**。

```python
# 可配置项：Memory 使用的 token 上限（本地模型需自定义 token 计数，否则会报 NotImplementedError）
MAX_TOKEN_LIMIT = 2048

import tiktoken
from langchain_openai import ChatOpenAI
from langchain_core.messages import BaseMessage

def _count_tokens_with_tiktoken(messages: list) -> int:
    """用 tiktoken 估算消息列表的 token 数（兼容本地/任意模型）"""
    enc = tiktoken.get_encoding("cl100k_base")
    n = 0
    for m in messages:
        n += 3  # 每条约 3 token 的 role/格式开销
        content = getattr(m, "content", None) or ""
        if isinstance(content, str):
            n += len(enc.encode(content))
        elif isinstance(content, list):
            for part in content:
                if isinstance(part, dict) and part.get("type") == "text":
                    n += len(enc.encode(part.get("text", "")))
                elif isinstance(part, str):
                    n += len(enc.encode(part))
    return n

class ChatOpenAIWithTokenCount(ChatOpenAI):
    """支持本地模型：用 tiktoken 实现 get_num_tokens_from_messages"""
    def get_num_tokens_from_messages(self, messages: list, **kwargs) -> int:
        return _count_tokens_with_tiktoken(messages)
```

```python
# 1. ConversationBufferWindowMemory：只保留最近 k 轮对话
from langchain.memory import ConversationBufferWindowMemory
from langchain.chains import ConversationChain
from langchain_openai import ChatOpenAI

LM_STUDIO_BASE = "http://127.0.0.1:1234/v1"
llm = ChatOpenAI(base_url=LM_STUDIO_BASE, api_key="lm-studio", temperature=0, model="local")

memory_window = ConversationBufferWindowMemory(k=2)  # 只保留最近 2 轮（2 个 human + 2 个 ai）
chain_window = ConversationChain(llm=llm, memory=memory_window, verbose=True)

chain_window.predict(input="我叫小明")
chain_window.predict(input="我喜欢跑步")
chain_window.predict(input="我也很喜欢散步")
chain_window.predict(input="我叫什么名字？")  # 若 k 太小，可能已经“忘掉”名字
```

```python
# 2. ConversationSummaryBufferMemory：旧对话压缩成摘要，只发「摘要 + 最近几轮」
# 需先运行上一个 cell，得到 MAX_TOKEN_LIMIT 和 ChatOpenAIWithTokenCount
from langchain.memory import ConversationSummaryBufferMemory
from langchain.chains import ConversationChain

LM_STUDIO_BASE = "http://127.0.0.1:1234/v1"
llm = ChatOpenAIWithTokenCount(base_url=LM_STUDIO_BASE, api_key="lm-studio", temperature=0, model="local")

# 超过 max_token_limit 的旧内容会被 summarizer 压缩成摘要；token 数使用上方可配置项
memory_summary = ConversationSummaryBufferMemory(llm=llm, max_token_limit=MAX_TOKEN_LIMIT)
chain_summary = ConversationChain(llm=llm, memory=memory_summary, verbose=True)

chain_summary.predict(input="我是产品经理，负责 APP 设计")
chain_summary.predict(input="我们团队有 5 个人，在上海")
chain_summary.predict(input="刚才我说了我的角色和团队情况，请简要复述一下")
```

```python
# 3. ConversationTokenBufferMemory：按 token 数截断，只保留最近 N 个 token
# 需先运行配置 cell，得到 MAX_TOKEN_LIMIT 和 ChatOpenAIWithTokenCount
from langchain.memory import ConversationTokenBufferMemory
from langchain.chains import ConversationChain

LM_STUDIO_BASE = "http://127.0.0.1:1234/v1"
llm = ChatOpenAIWithTokenCount(base_url=LM_STUDIO_BASE, api_key="lm-studio", temperature=0, model="local")

# 只保留最近 MAX_TOKEN_LIMIT 个 token（使用上方可配置项，默认 2048）
memory_token = ConversationTokenBufferMemory(llm=llm, max_token_limit=MAX_TOKEN_LIMIT)
chain_token = ConversationChain(llm=llm, memory=memory_token, verbose=True)

chain_token.predict(input="今天天气不错")
chain_token.predict(input="我想去公园散步")
chain_token.predict(input="我们刚才在聊什么？")
```

```python
import os
from langchain.chains import LLMChain, SimpleSequentialChain
from langchain import PromptTemplate
from langchain_openai import ChatOpenAI


LM_STUDIO_BASE = "http://127.0.0.1:1234/v1"
llm = ChatOpenAI(
    base_url=LM_STUDIO_BASE,
    api_key="lm-studio",  # 本地服务可不校验，任意非空即可
    temperature=0,
    model="local",        # 若 LM Studio 要求指定模型名，可改为你在 LM Studio 里加载的模型名
    request_timeout=120,  # 避免请求一直挂起：120 秒无响应则报错，可酌情调大
    max_tokens=1024,      # 限制单次生成长度，防止本地模型陷入“烤鸡翅、烤鸡腿”式无限重复
)

cate_template = "你现在是一位美食博主，需要根据用户输入的城市：{city} ，给出对应的美食推荐"
cate_prompt = PromptTemplate(template = cate_template, input_variables = ['city'])
# 第一个链
cate_chain = LLMChain(llm = llm, prompt= cate_prompt)
introduction_template ="你现在是一位美食博上，而要根据用户输入的美食信息： {cate} ，给出对应的美食简单介绍"
introduction_prompt = PromptTemplate(template = introduction_template,input_variables= ['cate'])
introduction_chain = LLMChain(llm = llm, prompt= introduction_prompt)
# 顺序链
result_chain = SimpleSequentialChain(chains = [cate_chain, introduction_chain], verbose= True)
# 这里看起来是把上面的输入框死，先生成prompt，再执行模型
# 第二个链
result_chain.run("成都")
```
