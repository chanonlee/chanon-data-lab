```python
import re

def calculate(expression):
    """
    计算表达式，支持 + - * / ** () 和小数
    """
    try:
        # 使用 eval 时注意安全，这里我们用更安全的解析方式
        # 但为了简单，这里使用 eval，实际项目中应使用更安全的解析器
        # 为避免安全风险，这里使用 eval 但限定在基础运算符
        result = eval(expression.replace('**', '**'))
        return result
    except:
        return "错误：表达式无效或包含非法字符"

def main():
    print("欢迎使用简单计算器！")
    print("支持运算：+ - * / ** （括号）")
    print("输入 'quit' 退出")
    
    while True:
        user_input = input("\n请输入表达式（或输入 'quit' 退出）: ").strip()
        
        if user_input.lower() == 'quit':
            print("再见！")
            break
        elif not user_input:
            print("输入为空，请输入有效表达式。")
            continue
            
        try:
            # 检查是否是数字或简单运算符
            if re.match(r'^[0-9]+\.?[0-9]*$', user_input):
                print(f"结果: {float(user_input)}")
            else:
                # 尝试计算表达式
                result = calculate(user_input)
                print(f"结果: {result}")
        except Exception as e:
            print(f"计算错误: {e}")

if __name__ == "__main__":
    main()
```
