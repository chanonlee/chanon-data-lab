```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
# 设置随机种子以确保结果可重现
np.random.seed(42)

# 创建模拟数据
n_samples = 100  # 样本数量

# 生成三个自变量（特征）
x1 = np.linspace(0, 10, n_samples)
x2 = np.random.rand(n_samples) * 5
x3 = np.random.randn(n_samples) * 2

# 添加随机噪声
noise = np.random.randn(n_samples) * 1.5

# 创建因变量（真实关系：y = 3 + 2*x1 - 1.5*x2 + 0.8*x3 + 噪声）
y = 3 + 2*x1 - 1.5*x2 + 0.8*x3 + noise

# 准备数据矩阵
# 添加一列1用于截距项
X = np.column_stack((np.ones(n_samples), x1, x2, x3))

print(X)
```

```python
# 手动计算回归系数
# 使用公式：β = (X^T * X)^{-1} * X^T * y
XT = X.T
XTX = XT.dot(X)
XTX_inv = np.linalg.inv(XTX)
XTy = XT.dot(y)
beta_manual = XTX_inv.dot(XTy)

print(beta_manual)
```

```python
# 打印手动计算的结果
print("手动计算的回归系数:")
print(f"截距 (β0): {beta_manual[0]:.6f}")
print(f"x1系数 (β1): {beta_manual[1]:.6f}")
print(f"x2系数 (β2): {beta_manual[2]:.6f}")
print(f"x3系数 (β3): {beta_manual[3]:.6f}")

```

```python
# 使用scikit-learn进行线性回归
model = LinearRegression()
model.fit(X[:, 1:], y)  # 排除截距列
# 打印scikit-learn的结果
print("\nscikit-learn计算的回归系数:")
print(f"截距 (β0): {model.intercept_:.6f}")
print(f"x1系数 (β1): {model.coef_[0]:.6f}")
print(f"x2系数 (β2): {model.coef_[1]:.6f}")
print(f"x3系数 (β3): {model.coef_[2]:.6f}")
```

```python
# 使用手动计算的系数进行预测
y_pred_manual = X.dot(beta_manual)
# 残差图
plt.subplot(2, 2, 4)
residuals = y - y_pred_manual
plt.scatter(y_pred_manual, residuals, alpha=0.7)
plt.axhline(y=0, color='r', linestyle='-')
plt.xlabel('prediction')
plt.ylabel('RSS')
plt.title('graph')
```

```python

# 3D可视化（展示x1, x2与y的关系）
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

# 绘制样本点
ax.scatter(x1, x2, y, c='b', marker='o', alpha=0.6, label='actual')

# 创建网格用于绘制回归平面
x1_grid, x2_grid = np.meshgrid(np.linspace(min(x1), max(x1), 10), 
                             np.linspace(min(x2), max(x2), 10))

# 取x3的平均值
x3_mean = np.mean(x3)
y_plane = beta_manual[0] + beta_manual[1]*x1_grid + beta_manual[2]*x2_grid + beta_manual[3]*x3_mean

# 绘制回归平面
ax.plot_surface(x1_grid, x2_grid, y_plane, color='r', alpha=0.3, label='regression plane')

ax.set_xlabel('x1')
ax.set_ylabel('x2')
ax.set_zlabel('y')
plt.title('multi linear regression plane (x3 as average)')
plt.legend()
plt.show()
```
