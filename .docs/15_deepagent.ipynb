### Deep Agents 依赖说明

PyPI 上的 **`deepagents` 要求 Python ≥ 3.11**。若 Jupyter 内核是 **3.9 / 3.10**，会出现 `No matching distribution found for deepagents`（不是包名写错）。

**对策：** 用 Python 3.11+ 重建 `.docs/.venv` 并选用该内核，例如在 `.docs` 目录执行：

```bash
python3.12 -m venv .venv
source .venv/bin/activate   # Windows: .venv\\Scripts\\activate
pip install -U pip ipykernel
pip install deepagents langchain-ollama langchain-openai
python -m ipykernel install --user --name=docs-deepagents --display-name="Python 3.12 (.docs)"
```

然后在 VS Code / Jupyter 里把内核换成 **Python 3.12 (.docs)**，再运行下一格的安装命令。

````python
import sys

if sys.version_info < (3, 11):
    raise RuntimeError(
        "deepagents 仅支持 Python 3.11+，当前为 "
        f"{sys.version.split()[0]}。请先按上一格说明用 3.11+ 重建 venv 并切换内核。"
    )

%pip install -qU pip
%pip install -qU deepagents langchain-ollama langchain-openai
````

````python
from deepagents import create_deep_agent
from llm_config import build_chat_openai

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

# 默认走 .env.local 的 LM Studio 配置，可通过 LM_PROVIDER 切换。
lm = build_chat_openai(temperature=0)

agent = create_deep_agent(
    model=lm,
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)

agent.invoke(
    {"messages": [{"role": "user", "content": "what is the weather in sf"}]}
)
````
