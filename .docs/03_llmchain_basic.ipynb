```python
# 导入：各层用到的类型。
# 整体自下而上：聊天模型 → 提示模板（槽位）→ LLMChain（编排）→ 传入变量调用。
from langchain_classic.chains import LLMChain
from langchain_core.prompts import PromptTemplate
from llm_config import build_chat_openai
```

```python
# 1. 模型层（LLM）：连接推理服务与采样参数；与提示文案无关，可单独替换或复用。
llm = build_chat_openai(temperature=1)
```

```python
# 2. 提示层（PromptTemplate）：模板长什么样、有哪些 {槽位}；input_variables 须与槽位名一致。
template = """
请写一首{adjective}风格的小诗，主题是{subject}。
要求：有画面感，带一点荒诞，最后一句要让人觉得好笑又认真。每首诗不要超过30个字。
"""
prompt = PromptTemplate(
    template=template,
    input_variables=["adjective", "subject"],
)
```

```python
# 3. 编排层（LLMChain）：把 PromptTemplate 与 LLM 绑成一条可重复执行的流水线；verbose=True 可打印格式化后的提示。
llm_chain = LLMChain(prompt=prompt, llm=llm, verbose=True)
```

```python
# 4. 调用层（业务入参）：业务只需要填槽位；链内完成「填模板 → 调模型 → 返回文本」。
# 用一个样例列表批量验证不同风格与主题。
samples = [
    ("赛博伤感", "凌晨两点还在保温的电饭煲"),
    ("一本正经", "坚信自己是皇帝的宿舍台灯"),
    ("过度热血", "便利店最后一根烤肠"),
    ("忧郁文艺", "被落在洗衣机里的单只袜子"),
    ("浮夸浪漫", "会发语音的蟑螂"),
    ("戏剧化", "被老板已读不回的打工人水杯"),
    ("中二史诗", "小区门口那台脾气很差的自动贩卖机"),
    ("假装成熟", "刚学会飞就想辞职的麻雀"),
]

for adjective, subject in samples:
    poem = llm_chain.predict(adjective=adjective, subject=subject)
    print(f"\n【{adjective}】{subject}")
    print(poem)
```
