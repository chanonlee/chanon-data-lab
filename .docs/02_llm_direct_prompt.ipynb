# LLM 直接文本提示（无 PromptTemplate）


```python
from llm_config import build_chat_openai

# =====================模型层=====================
# 此处可设置为本地模型或远程模型

llm = build_chat_openai(temperature=1)

# =====================提示层=====================
prompt_text = "Write a sad poem about ducks, gooses."

# =====================调用层=====================
response = llm.invoke(prompt_text)

# =====================输出层=====================
print(response.content)
```
